```python
# ---- CELL 1 ----
# Mount Drive, clone/pull repo, load config
# Run this every session

from google.colab import drive
drive.mount("/content/drive")

import os
import yaml
import sys
import subprocess

REPO_DIR = "/content/soccer-analytics"

if os.path.exists(REPO_DIR):
    subprocess.run(["git", "pull"], cwd=REPO_DIR, capture_output=True)
else:
    subprocess.run([
        "git", "clone",
        "https://github.com/Kareem-Ansari/soccer-analytics.git",
        REPO_DIR
    ], check=True)

os.chdir(REPO_DIR)
sys.path.insert(0, REPO_DIR)

with open("configs/config.yaml") as f:
    cfg = yaml.safe_load(f)

print("Setup complete")
print(f"  Repo: {REPO_DIR}")
```
